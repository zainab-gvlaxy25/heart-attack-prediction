<strong>EDA Exploratory Data Analysis     </strong>

```python
plt.figure(figsize=(12, 5))

# Histogram of Age
sns.histplot(df["Age"], bins=30, kde=True, color="darkblue")
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Count")
plt.show()
```

```python
plt.figure(figsize=(6, 4))
df["Outcome"].value_counts().plot(kind="pie", autopct="%1.1f%%", colors=["teal", "pink"])
plt.title("Heart Attack Outcome Distribution")
plt.ylabel("")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.countplot(x="FamilyHistory", hue="Outcome", data=df, palette="coolwarm")
plt.title("Family History vs. Heart Attack")
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.histplot(df["BMI"], kde=True, color="purple")
plt.title("BMI Distribution")
plt.xlabel("BMI")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.jointplot(data=df,x='Age',y='Income',kind='hex')
plt.title('The age wise income')
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.lineplot(x='Age',y='BMI',data=df,color='g')
plt.title('The age wise bmi')
plt.show()
```

```python
plt.figure(figsize=(6, 4))
sns.countplot(x="Gender", data=df, palette="pastel")
plt.title("Gender Distribution")
plt.show()
```

```python
plt.figure(figsize=(12, 5))
sns.countplot(x="Smoker", hue="Outcome", data=df, palette="pastel")
plt.title("Smoking Habit vs. Heart Attack Outcome")
plt.show()

plt.figure(figsize=(12, 5))
sns.boxplot(x="Outcome", y="BMI", data=df, palette="Set2")
plt.title("BMI vs. Heart Attack Outcome")
plt.show()
```
